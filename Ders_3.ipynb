```python
# Basit bir tahmin oyunu:
# Sayı gizli değeri sabit.
# tahmin hakkı 3 olsun.
# inputtan sayıyı girsin. Tahmin etsin.
# bildiyse bildin.
# input küçükse - daha büyük sayı ver.
# input büyükse - daha küçük sayı ver.
# haları biterse de hakkın bitti sayı ... du.
```

```python
gizli = 40
hak = 3

while hak > 0:
  tahmin = int(input(f"Tahiminin {hak} hakkın var: "))

  if tahmin == gizli:
    print("Bravo bildin.")
    break
  elif tahmin < gizli:
    print("Daha büyük bir sayı gir")
  else:
    print("Daha küçük bir sayı gir")

  hak -= 1 # hak = hak - 1 -

if hak == 0:
  print(f"Hakların bitti. Sayı: {gizli}")
```
