```python
import pandas as pd
import joblib
```

```python
#KAYDETTİĞİMİZ MODELİ İÇERİ AKTARDIK
yuklenen_model = joblib.load("/content/kredi_modeli.joblib")
print(yuklenen_model)
```

```python
#MODELİMİZİ ÖNCEKİ DATASETİMİZDEKİ X_test dataları ile test ettik. Ve modeli doğru kaydedebilmişmiyiz ona baktık.
X_test = pd.read_csv("/content/X_test.csv")
ilk = pd.read_csv("/content/ilk_tahminler.csv")

yeni_tahminler = yuklenen_model.predict(X_test) #Bu adımdan sonra tekrar skor da bakılabilir.
ayni_mi = (yeni_tahminler == ilk["tahmin"]).all()
print("tahminler aynı mi?", ayni_mi)
```

```python
basvuru = pd.DataFrame([{
    "yas": 34,
    "calisma_tipi": "Özel Sektör",
    "calisma_suresi_yil": 8,
    "ev_durumu": "Kira",
    "aylik_gelir": 65000,
    "kredi_skoru": 1450,
    "kredi_tutari": 250000,
    "vade_ay": 36,
}])

sonuc= yuklenen_model.predict(basvuru)[0]
print(sonuc)
olasiliklar = yuklenen_model.predict_proba(basvuru)[0]
print(olasiliklar)
```

Biz yukarıdaki örnek kişiye modelimiz kredisini onayladı. %948 ihtimalle onay olasılığı verdi.
