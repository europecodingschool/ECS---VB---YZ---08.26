```python
#Kütüphanelerimizi Kurma
import gradio as gr
import joblib
import pandas as pd
```

```python
#Kaydettiğimiz modeli içeri aktaralım.

model = joblib.load('/content/kredi_modeli.joblib')
```

```python
#tahmin fonksiyonu yazacağız python ile. Kullanıcı bilgilerini girip butona basınca bu fonksiyon çağıralacak.

def kredi_tahmin(yas, calisma_tipi, calisma_suresi_yil,ev_durumu, aylik_gelir,kredi_skoru,kredi_tutari,vade_ay):
  basvuru = pd.DataFrame([{
      "yas": yas,
      "calisma_tipi": calisma_tipi,
      "calisma_suresi_yil": calisma_suresi_yil,
      "ev_durumu": ev_durumu,
      "aylik_gelir": aylik_gelir,
      "kredi_skoru": kredi_skoru,
      "kredi_tutari": kredi_tutari,
      "vade_ay": vade_ay
  }])

  olasiliklar = model.predict_proba(basvuru)[0]
  return {"Kredin Reddedilir": olasiliklar[0],
          "Kredin Kabul Edilir": olasiliklar[1]}

```

```python
#FONKSİYONUN ÇALIŞIP ÇALIŞMADIĞINI ÖNYÜZE BAĞLAMADAN ÖNCE TEST ETTİK.
kredi_tahmin(34,"Özel Sektör",8,"Kira",65000,1450,250000,36)
```
