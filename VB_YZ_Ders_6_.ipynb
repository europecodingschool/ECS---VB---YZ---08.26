```python
df = df.drop_duplicates() # tekrarlayan satırları silmiş olduk. ve bir değişkene atadık. Yoksa ilerleyene adımlara taşımaz.
```

```python
df
```

```python
df.shape # silme işlemi sonrası tekrar dataframe'in şekline baktık. silinmiş doğruladık.
```

```python
df = df.reset_index(drop=True) # indexi resetledik. Silinen değerler sonrası.
```

```python
df.mean()
```

```python
df = df.fillna(df.mean()) #boş dataları sütun ortalamaları ile doldurduk.
```

```python
df
```

```python
df.isnull().sum()
```

```python
df["ortalama"] = (df["sinav1"] + df["sinav2"] + df["sinav3"]+ df["proje"]) / 4 # yeni bir sütun eklendi. Ortalama isminde. Sınavların ve projelerin toplamı / 4 olarak.
```

```python
df
```

```python
df["durum"] = np.where(df["ortalama"] >= 60, "GECTI", "KALDI") #yeni bir sütun ekliyoruz. Koşula bağlı. "ortalama " >=60 ise geçti yazıyor. değilse kaldı yazıyor.
```

```python
df.groupby("durum")[["calisma_saati","devamsizlik","ortalama"]].mean() #gruplama işlemi satırlar özelinde yaptık.
```
