```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/kanser_veriseti_detayli.csv")
```

```python
df.head() # data setimi incelediğimde ana amacım belirli bilgilerini aldığım hastanın kanser mi değil mi olduğunu tahmin edecek algoritmayı geliştirmek.
```

```python
plt.figure(figsize=(10,8)) # Değişkenler Arası Korelasyon Matrisi çizdirdik.
corr = df.corr()
sns.heatmap(corr, annot=True, cmap='viridis', linewidths=0.6)
plt.title('Detaylı Kanser Datası: Değişkenler Arası Korelasyon Matrisi')
plt.show()
```

```python
fig, axes = plt.subplots(1,2, figsize=(14,6))

sns.boxplot(data=df, x='Kanser_Mi', y='Hasta_Yasi', ax=axes[0])
axes[0].set_title('Kanser Durumuna Göre Hasta Yaşı')

sns.boxplot(data=df, x='Kanser_Mi', y='Hucre_Yogunlugu', ax=axes[1])
axes[1].set_title('Kanser Durumuna Göre Hücre Yoğunluğu')

plt.show()

```
