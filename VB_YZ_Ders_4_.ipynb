```python
#0. Öğrenciler ve Sınav Notları Analizi
#1. Veri oluşturalım - 2 değişken - 10 adet öğrenciler(isim manuel), 3 adet notlar (otomatik oluşturun)
#2. Veri Analizi - Öğrencilerin Not ortalamalarını yazdırın.
#3. Veri Analizi - Her sınavın ortalaması
#4. Veri Analizi - Sınıf geneli istatistik
#5. Veri Analizi - harf notu, geçti geçemedi gibi bir şey atayın.
```

```python
# 1. Veriyi Oluşturma

ogrenciler = np.array(['Ayşe','Fatma','Merve','Gizem','Yeşim','Hasan','Ahmet','Mehmet','Murat','İsmail'])
notlar = np.random.randint(30,100, size=(10,3))
notlar
```

```python
# 2. Veri Analizi - Öğrencilerin Not ortalamalarını yazdırın.

ortalamalar = notlar.mean(axis=1)
ortalamalar
```

```python
np.round(ortalamalar,)
```

```python
#3. Veri Analizi - Her sınavın ortalaması

sınav_ortalamalar = notlar.mean(axis=0)
np.round(sınav_ortalamalar,0)

```

```python
#4. Veri Analizi - Sınıf geneli istatistik

Genel_ortalama = notlar.mean()
Genel_ortalama
```

```python
np.round(Genel_ortalama,0)
```

```python
en_yuksek_not = notlar.max()
en_yuksek_not
```

```python
en_dusuk_not = notlar.min()
en_dusuk_not
```
