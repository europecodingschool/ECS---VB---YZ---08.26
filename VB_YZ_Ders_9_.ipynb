```python
#x ve y ayrımı train test ayrımı.
X = df[["Study Hours","Sleep Hours", "Attendance Rate", "Social Media Hours"]]
y = df["Exam Score"]
```

```python
X.head()
```

```python
y.head()
```

```python
#train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25, random_state=15)
```

```python
#Ödev
#Standardize edip sonrasında modele öğretelim. ve başarı parametrelerini hesaplayalım.
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression
regression =LinearRegression()
```

```python
regression.fit(X_train,y_train)
```

```python
#Pred
y_pred=regression.predict(X_test)
```

```python
y_pred
```

```python
#Performance Metrics
from sklearn.metrics import mean_absolute_error,mean_squared_error
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
rmse=np.sqrt(mse)
print("mse:", mse) #MSE (Mean Squared Error - Hata Kareler Ortalaması) - Aykırı değerleri (outliers) cezalandırmak için mükemmeldir.
print("mae:", mae) #MAE (Mean Absolute Error - Ortalama Mutlak Hata):modelimiz ortalama 3 puan saparak tahminleme yapıyor.
print("rmse:", rmse) #RMSE (Root Mean Squared Error - Kök Hata Kareler Ortalaması):MSE'nin karesini alarak şişirdiği birimi tekrar orijinal veri birimine (TL, $m^2$) geri döndürür.
                     #Büyük hatalara karşı MAE'den çok daha duyarlıdır.
```

```python
#Performance Metrics
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred) # Değeri 0 ile 1 (veya %0 ile %100) arasındadır.
print("r2 score:",score)
```
